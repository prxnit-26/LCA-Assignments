```python
# to check whether a string contains only specific characters {a-z, A-Z, 0-9}
import re 
s = input("enter string: ")
result = re.fullmatch(r"[a-zA-Z0-9]+", s)
if result:
    print("string contains only alphanumeric characters")
else:
    print("string does not contain ONLY alphanumeric characters")
```
